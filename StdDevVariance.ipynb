## Activity

Experiment with different parameters on the normal function, and see what effect it has on the shape of the distribution. How does that new shape relate to the standard deviation and variance?

```python
import numpy as np
import matplotlib.pyplot as plt

# 1) Baseline: mean=100, std=50
data = np.random.normal(100, 50, 10000)
plt.hist(data, bins=50)
plt.title("Baseline: mean=100, std=50")
plt.show()
print("Baseline -> std:", data.std(), "var:", data.var())

# 2) Shift mean only: mean=150, std=50
data = np.random.normal(150, 50, 10000)
plt.hist(data, bins=50)
plt.title("Shifted mean: mean=150, std=50")
plt.show()
print("Shifted mean -> std:", data.std(), "var:", data.var())

# 3) Smaller std: mean=100, std=10
data = np.random.normal(100, 10, 10000)
plt.hist(data, bins=50)
plt.title("Smaller spread: mean=100, std=10")
plt.show()
print("Small std -> std:", data.std(), "var:", data.var())

# 4) Larger std: mean=100, std=80
data = np.random.normal(100, 80, 10000)
plt.hist(data, bins=50)
plt.title("Larger spread: mean=100, std=80")
plt.show()
print("Large std -> std:", data.std(), "var:", data.var())

```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 200, 1000)

def normal_pdf(x, mu, sigma):
    return (1 / (sigma * np.sqrt(2*np.pi))) * np.exp(-0.5 * ((x - mu)/sigma)**2)

plt.figure(figsize=(8, 6))

# 1) Baseline: mean=100, std=50
plt.plot(x, normal_pdf(x, 100, 50), label='mean=100, σ=50')

# 2) Shift mean only: mean=150, std=50
plt.plot(x, normal_pdf(x, 150, 50), label='mean=150, σ=50')

# 3) Smaller std: mean=100, std=10
plt.plot(x, normal_pdf(x, 100, 10), label='mean=100, σ=10')

# 4) Larger std: mean=100, std=80
plt.plot(x, normal_pdf(x, 100, 80), label='mean=100, σ=80')

plt.legend()
plt.xlabel("x")
plt.ylabel("density")
plt.title("Normal PDFs with different means and standard deviations")
plt.show()

```
